```python
# Import required libraries
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
# Create the Music Streaming dataset
data = {
    "User_ID": [101, 102, 103, 104, 105, 106, 102, 108, 109, 110],
    "Song_Name": ["Song A", "Song B", "Song C", "Song D", "Song E",
                  "Song F", "Song B", "Song H", "Song I", "Song J"],
    "Genre": ["Hip-Hop", "Pop", "Rock", "Hip-Hop", None,
              "Pop", "Pop", "Jazz", "Rock", "Hip-Hop"],
    "Artist": ["Artist A", "Artist B", "Artist C", "Artist A",
               "Artist D", "Artist B", "Artist B", "Artist E",
               "Artist C", "Artist A"],
    "Age": [20, 22, np.nan, 24, 21, 23, 22, 25, 19, 26],
    "Listening_Time": [30, 45, 25, 60, 35, 50, 45, 20, 40, 55],
    "Stream_Count": [5, 8, 4, 10, 6, 9, 8, 3, 7, 11],
    "Subscription_Type": ["Premium", "Free", "Premium", "Free", "Premium",
                          "Free", "Free", "Premium", None, "Premium"]
}

df = pd.DataFrame(data)
print("===== ORIGINAL MUSIC STREAMING DATASET =====")
display(df)
print("\nDataset Shape:", df.shape)
```

```python
# Check missing values
print("\n===== MISSING VALUES IN DATASET =====")
print(df.isnull().sum())
```

```python
# Handle missing values
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Genre"] = df["Genre"].fillna(df["Genre"].mode()[0])
df["Subscription_Type"] = df["Subscription_Type"].fillna(
    df["Subscription_Type"].mode()[0]
)

print("\n===== DATASET AFTER HANDLING MISSING VALUES =====")
display(df)
print("\n===== MISSING VALUES AFTER HANDLING =====")
print(df.isnull().sum())
```

```python
# Check and remove duplicate records
print("\n===== DUPLICATE RECORDS =====")
print("Number of duplicate rows:", df.duplicated().sum())

df = df.drop_duplicates()
print("\n===== DATASET AFTER REMOVING DUPLICATES =====")
display(df)
print("Duplicates after removal:", df.duplicated().sum())
print("Dataset Shape:", df.shape)
```

```python
# Encode categorical columns
genre_encoder = LabelEncoder()
df["Genre"] = genre_encoder.fit_transform(df["Genre"])

subscription_encoder = LabelEncoder()
df["Subscription_Type"] = subscription_encoder.fit_transform(
    df["Subscription_Type"]
)

print("\n===== ENCODED DATASET =====")
display(df)
print("Genre Encoding:", dict(zip(genre_encoder.classes_,
                   genre_encoder.transform(genre_encoder.classes_))))
print("Subscription Type Encoding:", dict(zip(
    subscription_encoder.classes_,
    subscription_encoder.transform(subscription_encoder.classes_)
)))
```

```python
# Feature scaling
numerical_columns = ["Age", "Listening_Time", "Stream_Count"]
scaler = StandardScaler()
df[numerical_columns] = scaler.fit_transform(df[numerical_columns])

print("\n===== DATASET AFTER FEATURE SCALING =====")
display(df)
print("\n===== SCALED NUMERICAL COLUMNS =====")
display(df[numerical_columns])
```

```python
# Final verification
print("\n===== FINAL DATASET VERIFICATION =====")
print("\nMissing Values:")
print(df.isnull().sum())
print("\nDuplicate Rows:")
print(df.duplicated().sum())
print("\nFinal Dataset Shape:", df.shape)
```

```python
# Save preprocessed dataset
output_file = "Preprocessed_Music_Streaming.xlsx"
df.to_excel(output_file, index=False)
print("Preprocessed dataset saved successfully!")
print("File name:", output_file)
```
