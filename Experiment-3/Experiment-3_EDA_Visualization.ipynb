This notebook performs EDA on an illustrative Music Streaming System dataset. The report describes 15 songs and six attributes: Song, Artist, Genre, Listening_Time_Minutes, Stream_Count, and User_Rating. The exact original row-level dataset is not included in the report, so the sample rows below are recreated for demonstration.

```python
# 1. Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print('All libraries imported successfully!')
```

```python
# 2. Create the sample Music Streaming dataset
# Illustrative data created for this practical (not actual platform data).
data = {
    "Song": [f"Song {i}" for i in range(1, 16)],
    "Artist": ["Artist A", "Artist B", "Artist C", "Artist A", "Artist D",
               "Artist B", "Artist E", "Artist C", "Artist A", "Artist D",
               "Artist B", "Artist E", "Artist C", "Artist D", "Artist A"],
    "Genre": ["Pop", "Hip-Hop", "Rock", "Pop", "Jazz", "Hip-Hop", "Pop",
              "Rock", "Jazz", "Pop", "Hip-Hop", "Rock", "Pop", "Jazz", "Hip-Hop"],
    "Listening_Time_Minutes": [20, 25, 25, 30, 30, 30, 35, 35, 40, 40, 45, 45, 50, 55, 70],
    "Stream_Count": [900, 1100, 1200, 1500, 1600, 1800, 2000, 2200, 2200, 2500, 2700, 2800, 3200, 3500, 4500],
    "User_Rating": [3.2, 3.5, 3.6, 3.8, 4.0, 3.9, 4.1, 4.0, 4.2, 4.3, 4.4, 4.2, 4.5, 4.6, 4.8]
}

df = pd.DataFrame(data)
print("===== MUSIC STREAMING DATASET =====")
display(df)
print("\nDataset Shape:", df.shape)
```

```python
# 3. Inspect dataset structure and quality
print("===== FIRST FIVE ROWS =====")
display(df.head())
print("\n===== DATASET INFORMATION =====")
df.info()
print("\n===== MISSING VALUES =====")
print(df.isnull().sum())
print("\n===== DUPLICATE RECORDS =====")
print("Number of duplicate rows:", df.duplicated().sum())
print("\n===== DESCRIPTIVE STATISTICS =====")
display(df.describe(include='all'))
```

```python
# 4. Measures of central tendency
for column in ['Listening_Time_Minutes', 'Stream_Count']:
    print(f"===== {column} =====")
    print("Mean:", df[column].mean())
    print("Median:", df[column].median())
    print("Mode:", df[column].mode().tolist())
    print()
```

```python
# 5. Measures of dispersion
for column in ['Listening_Time_Minutes', 'Stream_Count']:
    values = df[column]
    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)
    print(f"===== {column} =====")
    print("Standard Deviation:", values.std())
    print("Variance:", values.var())
    print("Range:", values.max() - values.min())
    print("IQR:", q3 - q1)
    print()
```

```python
# 6. Correlation analysis
numeric_df = df[['Listening_Time_Minutes', 'Stream_Count', 'User_Rating']]
correlation_matrix = numeric_df.corr()
print("===== CORRELATION MATRIX =====")
display(correlation_matrix)
```

```python
# 7. Histogram of Listening Time
plt.figure(figsize=(8, 5))
plt.hist(df['Listening_Time_Minutes'], bins=6, edgecolor='black')
plt.title('Distribution of Song Listening Time')
plt.xlabel('Listening Time (Minutes)')
plt.ylabel('Number of Songs')
plt.grid(axis='y', alpha=0.3)
plt.show()
```

```python
# 8. Boxplot of Stream Count
plt.figure(figsize=(8, 5))
sns.boxplot(x=df['Stream_Count'])
plt.title('Boxplot of Song Stream Counts')
plt.xlabel('Number of Streams')
plt.show()
```

```python
# 9. Correlation Heatmap
plt.figure(figsize=(7, 5))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Heatmap of Music Streaming Data')
plt.show()
```

```python
# 10. Scatter Plot: Listening Time vs Stream Count
plt.figure(figsize=(8, 5))
plt.scatter(df['Listening_Time_Minutes'], df['Stream_Count'], edgecolors='black')
plt.title('Listening Time vs Stream Count')
plt.xlabel('Listening Time (Minutes)')
plt.ylabel('Number of Streams')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
# 11. Final summary and export dataset
print("===== FINAL DATASET SUMMARY =====")
print("Total Songs:", len(df))
print("Missing Values:", df.isnull().sum().sum())
print("Duplicate Rows:", df.duplicated().sum())
print("\nCorrelation Matrix:")
display(correlation_matrix)

df.to_csv('Music_Streaming_EDA_Dataset.csv', index=False)
print('Dataset exported as Music_Streaming_EDA_Dataset.csv')
```
