```python
# 1. Import required libraries
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

print('Libraries imported successfully!')
```

```python
# 2. Upload the CSV file in Google Colab
from google.colab import files
uploaded = files.upload()

# Make sure the uploaded file is named Music_Streaming_Subscriber_Churn.csv
file_name = 'Music_Streaming_Subscriber_Churn.csv'
if file_name not in uploaded:
    raise FileNotFoundError(f'Please upload {file_name}')

df = pd.read_csv(file_name)
print('Dataset loaded successfully!')
```

```python
# 3. Inspect the dataset
print("First 5 rows of the dataset:")
display(df.head())
print("\nDataset Shape:", df.shape)
print("\nColumn Names:", df.columns.tolist())
print("\nMissing Values:")
print(df.isnull().sum())
```

```python
# 4. Encode categorical columns and prepare features/target
data = df.copy()
categorical_cols = ['subscription_plan', 'favorite_genre', 'device', 'country']

for col in categorical_cols:
    encoder = LabelEncoder()
    data[col] = encoder.fit_transform(data[col].astype(str))

target_encoder = LabelEncoder()
data['churn_status'] = target_encoder.fit_transform(data['churn_status'].astype(str))

X = data.drop(columns=['user_id', 'churn_status'])
y = data['churn_status']

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)

print('Preprocessing completed successfully!')
print('Training samples:', len(X_train))
print('Testing samples:', len(X_test))
print('Number of features:', X_train.shape[1])
print('Target classes:', np.unique(y_train))
print('Target label mapping:', dict(zip(target_encoder.classes_, target_encoder.transform(target_encoder.classes_))))
```

```python
# 5. Inbuilt Decision Tree Classifier (Gini)
dt_model = DecisionTreeClassifier(
    criterion='gini',
    max_depth=5,
    random_state=42
)
dt_model.fit(X_train, y_train)
dt_predictions = dt_model.predict(X_test)

print('INBUILT DECISION TREE CLASSIFIER')
print('Accuracy:', round(accuracy_score(y_test, dt_predictions) * 100, 2), '%')
print('\nClassification Report:')
print(classification_report(y_test, dt_predictions, zero_division=0))
print('Confusion Matrix:')
print(confusion_matrix(y_test, dt_predictions))
```

```python
# 6. User-defined Decision Tree Classifier (Entropy)
def decision_tree_classifier(X_train, X_test, y_train, y_test):
    model = DecisionTreeClassifier(
        criterion='entropy',
        max_depth=5,
        random_state=42
    )
    model.fit(X_train, y_train)
    predictions = model.predict(X_test)

    print('USER-DEFINED DECISION TREE CLASSIFIER')
    print('Accuracy:', round(accuracy_score(y_test, predictions) * 100, 2), '%')
    print('\nClassification Report:')
    print(classification_report(y_test, predictions, zero_division=0))
    print('Confusion Matrix:')
    print(confusion_matrix(y_test, predictions))
    return model, predictions

custom_dt_model, custom_dt_predictions = decision_tree_classifier(
    X_train, X_test, y_train, y_test
)
```

```python
# 7. Inbuilt Naïve Bayes Classifier (GaussianNB)
nb_model = GaussianNB()
nb_model.fit(X_train, y_train)
nb_predictions = nb_model.predict(X_test)

print('INBUILT NAIVE BAYES CLASSIFIER')
print('Accuracy:', round(accuracy_score(y_test, nb_predictions) * 100, 2), '%')
print('\nClassification Report:')
print(classification_report(y_test, nb_predictions, zero_division=0))
print('Confusion Matrix:')
print(confusion_matrix(y_test, nb_predictions))
```

```python
# 8. User-defined Naïve Bayes Classifier
def naive_bayes_classifier(X_train, X_test, y_train, y_test):
    model = GaussianNB()
    model.fit(X_train, y_train)
    predictions = model.predict(X_test)

    print('USER-DEFINED NAIVE BAYES CLASSIFIER')
    print('Accuracy:', round(accuracy_score(y_test, predictions) * 100, 2), '%')
    print('\nClassification Report:')
    print(classification_report(y_test, predictions, zero_division=0))
    print('Confusion Matrix:')
    print(confusion_matrix(y_test, predictions))
    return model, predictions

custom_nb_model, custom_nb_predictions = naive_bayes_classifier(
    X_train, X_test, y_train, y_test
)
```

```python
# 9. Compare all four classifiers
results = pd.DataFrame({
    'Classifier': [
        'Inbuilt Decision Tree',
        'User-Defined Decision Tree',
        'Inbuilt Naive Bayes',
        'User-Defined Naive Bayes'
    ],
    'Accuracy (%)': [
        round(accuracy_score(y_test, dt_predictions) * 100, 2),
        round(accuracy_score(y_test, custom_dt_predictions) * 100, 2),
        round(accuracy_score(y_test, nb_predictions) * 100, 2),
        round(accuracy_score(y_test, custom_nb_predictions) * 100, 2)
    ]
})

print('CLASSIFIER COMPARISON')
display(results)
best_model = results.loc[results['Accuracy (%)'].idxmax()]
print('Best Classifier:', best_model['Classifier'])
print('Best Accuracy:', best_model['Accuracy (%)'], '%')
```
