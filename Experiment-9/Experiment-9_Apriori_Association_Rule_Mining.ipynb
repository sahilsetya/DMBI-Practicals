```python
# 1. Import required libraries
import pandas as pd
import matplotlib.pyplot as plt
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
# 2. Load Dataset
from google.colab import files
uploaded = files.upload()

df = pd.read_csv("Music_Streaming_Subscriber_Churn.csv")

print("\nDataset Loaded Successfully")
print("--------------------------------")
print("Dataset Shape:", df.shape)
print("\nFirst 5 Records:")
display(df.head())
```

```python
# 3. Select Categorical Attributes
categorical_columns = [
    "subscription_plan",
    "favorite_genre",
    "device",
    "country",
    "churn_status"
]

items_df = df[categorical_columns].copy()
items_df = items_df.fillna("Unknown")
```

```python
# 4. Convert Data into Basket Format
basket = pd.get_dummies(
    items_df,
    prefix_sep="=",
    dtype=bool
)

print("\nTransaction Data Created")
print("--------------------------------")
print("Total Transactions:", len(basket))
print("Total Items:", len(basket.columns))
```

```python
# 5. Apply Apriori Algorithm
frequent_itemsets = apriori(
    basket,
    min_support=0.05,
    use_colnames=True,
    max_len=3
)

frequent_itemsets = frequent_itemsets.sort_values(
    by="support",
    ascending=False
)

print("\nFrequent Itemsets Generated")
print("--------------------------------")
print("Total Frequent Itemsets:", len(frequent_itemsets))
display(frequent_itemsets.head(10))
```

```python
# 6. Generate Association Rules
rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.50
)

rules = rules.sort_values(
    by="lift",
    ascending=False
)

strong_rules = rules[rules["lift"] > 1]

print("\nAssociation Rules Generated")
print("--------------------------------")
print("Total Rules:", len(rules))
print("\nStrong Rules (Lift > 1)")
print("--------------------------------")
print("Total Strong Rules:", len(strong_rules))
```

```python
# 7. Save Output CSV Files
frequent_itemsets.to_csv("frequent_itemsets.csv", index=False)
rules.to_csv("association_rules.csv", index=False)
strong_rules.to_csv("strong_association_rules.csv", index=False)

print("\nCSV Files Saved Successfully")
```

```python
# 8. Prepare Top 10 Frequent Itemsets for Visualization
top_items = frequent_itemsets.head(10).copy()
top_items["itemsets"] = top_items["itemsets"].apply(
    lambda x: ", ".join(sorted(list(x)))
)
```

```python
# 9. Visualize Frequent Itemsets and Association Rules
fig, axes = plt.subplots(1, 2, figsize=(18, 7))

# Frequent Itemsets Graph
axes[0].bar(
    range(len(top_items)),
    top_items["support"]
)
axes[0].set_title("Top 10 Frequent Itemsets")
axes[0].set_xlabel("Itemsets")
axes[0].set_ylabel("Support")
axes[0].set_xticks(range(len(top_items)))
axes[0].set_xticklabels(
    top_items["itemsets"],
    rotation=45,
    ha="right",
    fontsize=8
)

# Association Rules Graph
scatter = axes[1].scatter(
    rules["support"],
    rules["confidence"],
    c=rules["lift"],
    cmap="viridis",
    s=60
)
axes[1].set_title("Association Rules")
axes[1].set_xlabel("Support")
axes[1].set_ylabel("Confidence")

fig.colorbar(
    scatter,
    ax=axes[1],
    label="Lift"
)

fig.suptitle(
    "Experiment 9 - Apriori Association Rule Mining",
    fontsize=14
)
plt.tight_layout()
plt.savefig("Experiment9_Graphs.png", dpi=300, bbox_inches="tight")
plt.show()
plt.close()

print("\nGraph Saved Successfully")
print("File Name: Experiment9_Graphs.png")
```
